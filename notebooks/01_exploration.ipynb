three candidates for the excluded segment

```python
def segment_share(segment_df, total_len):
    return len(segment_df) / total_len * 100


def segment_churn_rate(segment_df):
    return len(segment_df[segment_df["Churn"] == "Yes"]) / len(segment_df) * 100
```

```python
df.groupby("Contract").count()
```

```python
df.groupby("PaymentMethod").count()
```

```python
df.groupby("tenure").count()
```

```python
total_len = len(df)

for col in ["Contract", "PaymentMethod"]:
    summary = df.groupby(col).apply(
        lambda g: pd.Series({
            "pct_of_total": segment_share(g, total_len),
            "churn_rate": segment_churn_rate(g)
        }),
        include_groups=False,
    )
    print(f"\n{col}")
    print(summary.round(2))
```

```python
tenure_bins = [0, 18, 36, 60, 72]
tenure_labels = ["0-18", "18-36", "36-60", "60-72"]

df["tenure_group"] = pd.cut(df["tenure"], bins=tenure_bins, labels=tenure_labels, include_lowest=True)

tenure_summary = df.groupby("tenure_group", observed=True).apply(
    lambda g: pd.Series({
        "pct_of_total": segment_share(g, total_len),
        "churn_rate": segment_churn_rate(g),
    }),
    include_groups=False,
)
print("tenure_group")
print(tenure_summary.round(2))
```
